```python
data=pd.read_csv("/content/diabetes.csv")
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

```python
x=data.drop(['Outcome'],axis=1)
y=data['Outcome']
```

```python
x.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
x_test.head()
```

```python
x_train.shape[1]
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
model=keras.Sequential([
    layers.Input(shape=(x_train_scaled.shape[1],),name='input_layer'), #input_layer has 8 neurons
    layers.Dense(32,activation='relu',name='hidden_layer_1'), #hidden_layer_1 has 32 neurons
    layers.Dense(16,activation='relu',name='hidden_layer_2'), #hidden_layer_2 has 16 neurons
    layers.Dense(1,activation='sigmoid',name='output_layer') #output_layer has just 1 neuron since it is a binary classifier
])
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
model.fit(x_train_scaled,y_train,epochs=8,batch_size=16)
```

```python
y_pred_prob=model.predict(x_test_scaled)
```

```python
print(y_pred_prob)
```

```python
y_pred = (y_pred_prob > 0.5).astype(int)
```

```python
print(y_pred)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
accuracy = accuracy_score(y_test, y_pred)
print(accuracy)
```

```python
cm = confusion_matrix(y_test, y_pred)
print(cm)
```
